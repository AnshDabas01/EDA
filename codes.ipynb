```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
file_path = "retail_customer_sales_dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include='all').T)
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

categorical_columns = df.select_dtypes(
    include=['object', 'category', 'string']
).columns.tolist()

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)
```

```python
missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_report = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage
})

missing_report = missing_report.sort_values(
    by='Missing Values', ascending=False
)

display(missing_report)
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)

print("\nDuplicate Records:")
display(df[df.duplicated(keep=False)].head(10))
```

```python
if 'Transaction_ID' in df.columns:
    print(
        "Duplicate Transaction IDs:",
        df['Transaction_ID'].duplicated().sum()
    )
```

```python
# Create a copy of the original dataset

df_clean = df.copy()

print("Clean dataset copy created successfully!")
```

```python
# Remove extra spaces from categorical columns

for col in categorical_columns:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

# Standardize Gender
if 'Gender' in df_clean.columns:
    df_clean['Gender'] = df_clean['Gender'].replace({
        'm': 'Male',
        'M': 'Male',
        'male': 'Male',
        'MALE': 'Male',
        'f': 'Female',
        'F': 'Female',
        'female': 'Female',
        'FEMALE': 'Female'
    })

# Standardize Region
if 'Region' in df_clean.columns:
    df_clean['Region'] = df_clean['Region'].replace({
        'north': 'North',
        'NORTH': 'North',
        'south': 'South',
        'SOUTH': 'South',
        'east': 'East',
        'EAST': 'East',
        'west': 'West',
        'WEST': 'West'
    })

# Standardize other categorical columns
standardize_columns = [
    'Customer_Segment',
    'Product_Category',
    'Product',
    'Payment_Method',
    'Delivery_Mode',
    'Satisfaction_Level',
    'Returned'
]

for col in standardize_columns:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].apply(
            lambda x: x.title().strip() if isinstance(x, str) else x
        )

print("Categorical values standardized successfully!")
```

```python
# Invalid ages
if 'Age' in df_clean.columns:
    df_clean.loc[
        (df_clean['Age'] < 0) | (df_clean['Age'] > 100),
        'Age'
    ] = np.nan

```

```python


# Invalid discounts
if 'Discount' in df_clean.columns:
    df_clean.loc[
        (df_clean['Discount'] < 0) |
        (df_clean['Discount'] > 0.5),
        'Discount'
    ] = np.nan

```

```python


# Invalid quantities
if 'Quantity' in df_clean.columns:
    df_clean.loc[
        df_clean['Quantity'] <= 0,
        'Quantity'
    ] = np.nan
```

```python


# Negative unit prices
if 'Unit_Price' in df_clean.columns:
    df_clean.loc[
        df_clean['Unit_Price'] < 0,
        'Unit_Price'
    ] = np.nan

print("Invalid numerical values handled.")
```

```python
# Define date columns
date_columns = ['Order_Date', 'Delivery_Date']

print("Date columns defined successfully!")
```

```python
for col in date_columns:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(
            df_clean[col],
            format='%Y-%m-%d',
            errors='coerce'
        )

# Handle delivery dates earlier than order dates
invalid_dates = (
    df_clean['Delivery_Date'] < df_clean['Order_Date']
)

df_clean.loc[invalid_dates, 'Delivery_Date'] = pd.NaT

print("Date columns converted successfully.")

print(df_clean[['Order_Date', 'Delivery_Date']].dtypes)
```

```python
# Numerical columns for median imputation
numeric_impute_columns = [
    'Age', 'Income', 'Quantity',
    'Unit_Price', 'Discount',
    'Sales', 'Cost', 'Profit'
]

for col in numeric_impute_columns:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )

# Categorical columns for mode imputation
categorical_impute_columns = [
    'Gender', 'Region', 'State', 'City',
    'Customer_Segment', 'Product_Category',
    'Product', 'Payment_Method',
    'Delivery_Mode', 'Satisfaction_Level',
    'Returned'
]

for col in categorical_impute_columns:
    if col in df_clean.columns:
        mode_values = df_clean[col].mode(dropna=True)

        if not mode_values.empty:
            df_clean[col] = df_clean[col].fillna(
                mode_values.iloc[0]
            )

print("Missing value treatment completed.")
```

```python
remaining_missing = pd.DataFrame({
    'Missing Count': df_clean.isnull().sum(),
    'Missing Percentage':
        (df_clean.isnull().sum() / len(df_clean)) * 100
})

display(remaining_missing)
```

```python
key_numerical_columns = [
    'Age', 'Income', 'Quantity', 'Unit_Price',
    'Discount', 'Sales', 'Cost', 'Profit'
]

available_columns = [
    col for col in key_numerical_columns
    if col in df_clean.columns
]

statistics = df_clean[available_columns].describe().T

statistics['Mode'] = df_clean[available_columns].mode().iloc[0]

statistics['Variance'] = df_clean[available_columns].var()

statistics['IQR'] = (
    df_clean[available_columns].quantile(0.75) -
    df_clean[available_columns].quantile(0.25)
)

statistics['Skewness'] = df_clean[available_columns].skew()

display(statistics)
```

```python
selected_columns = [
    'Age', 'Income', 'Sales', 'Profit'
]

for col in selected_columns:
    if col in df_clean.columns:

        plt.figure(figsize=(8, 5))

        sns.histplot(
            df_clean[col].dropna(),
            kde=True,
            bins=30
        )

        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')

        plt.show()

        print(
            f"{col} Skewness:",
            df_clean[col].skew()
        )
```

```python
categorical_analysis = [
    'Region',
    'Product_Category',
    'Customer_Segment'
]

for col in categorical_analysis:
    if col in df_clean.columns:

        print(f"\nFrequency of {col}:")

        frequency = df_clean[col].value_counts()

        proportion = df_clean[col].value_counts(
            normalize=True
        ) * 100

        result = pd.DataFrame({
            'Frequency': frequency,
            'Percentage': proportion
        })

        display(result)

        plt.figure(figsize=(8, 5))

        sns.countplot(
            data=df_clean,
            x=col,
            order=df_clean[col].value_counts().index
        )

        plt.title(f'{col} Distribution')
        plt.xticks(rotation=30)
        plt.show()
```

```python
outlier_columns = ['Sales', 'Profit', 'Age', 'Income']

outlier_summary = []

for col in outlier_columns:

    if col in df_clean.columns:

        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR

        outliers = df_clean[
            (df_clean[col] < lower_limit) |
            (df_clean[col] > upper_limit)
        ]

        outlier_summary.append({
            'Column': col,
            'Lower Limit': lower_limit,
            'Upper Limit': upper_limit,
            'Outlier Count': len(outliers),
            'Outlier Percentage':
                (len(outliers) / len(df_clean)) * 100
        })

        print(f"\nOutliers in {col}:")
        display(outliers[[col]].head(10))

        plt.figure(figsize=(8, 4))

        sns.boxplot(x=df_clean[col])

        plt.title(f'Outlier Detection - {col}')
        plt.show()

outlier_report = pd.DataFrame(outlier_summary)

display(outlier_report)
```

```python
if all(col in df_clean.columns for col in ['Profit', 'Sales']):
    
    df_clean['Profit_Margin'] = np.where(
        df_clean['Sales'] != 0,
        (df_clean['Profit'] / df_clean['Sales']) * 100,
        np.nan
    )

    print("Profit Margin Created:")
    display(df_clean[['Sales', 'Profit', 'Profit_Margin']].head())
```

```python
if all(col in df_clean.columns for col in date_columns):
    
    df_clean['Delivery_Days'] = (
        df_clean['Delivery_Date'] -
        df_clean['Order_Date']
    ).dt.days

    print("Delivery Days Created:")
    display(df_clean[
        ['Order_Date', 'Delivery_Date', 'Delivery_Days']
    ].head())
```

```python
if all(col in df_clean.columns for col in ['Customer_ID', 'Sales']):
    
    df_clean['Revenue_per_Customer'] = (
        df_clean.groupby('Customer_ID')['Sales']
        .transform('sum')
    )

    print("Revenue per Customer Created:")
    display(df_clean[
        ['Customer_ID', 'Sales', 'Revenue_per_Customer']
    ].head())
```

```python
if all(col in df_clean.columns for col in ['Region', 'Sales']):
    
    region_sales = df_clean.groupby('Region')['Sales'].agg(
        ['sum', 'mean', 'count']
    ).sort_values(by='sum', ascending=False)

    display(region_sales)

    print("Region with Highest Sales:")
    print(region_sales['sum'].idxmax())
```

```python
if all(col in df_clean.columns for col in ['Product_Category', 'Profit']):
    
    category_profit = df_clean.groupby(
        'Product_Category'
    )['Profit'].agg(['sum', 'mean']).sort_values(
        by='sum', ascending=False
    )

    display(category_profit)

    print("Most Profitable Category:")
    print(category_profit['sum'].idxmax())
```

```python
if all(col in df_clean.columns for col in ['Customer_Segment', 'Sales']):
    
    segment_sales = df_clean.groupby(
        'Customer_Segment'
    )['Sales'].agg(['sum', 'mean', 'count']).sort_values(
        by='sum', ascending=False
    )

    display(segment_sales)

    print("Highest Revenue Customer Segment:")
    print(segment_sales['sum'].idxmax())
```

```python
if all(col in df_clean.columns for col in ['Region', 'Profit_Margin']):
    
    region_margin = df_clean.groupby(
        'Region'
    )['Profit_Margin'].mean().sort_values(
        ascending=False
    )

    display(region_margin)

    print("Highest Average Profit Margin:")
    print(region_margin.idxmax())
```

```python
if all(col in df_clean.columns for col in ['Product_Category', 'Profit_Margin']):
    
    category_margin = df_clean.groupby(
        'Product_Category'
    )['Profit_Margin'].mean().sort_values()

    display(category_margin)

    print("Lowest Profit Margin Category:")
    print(category_margin.idxmin())
```

```python
if 'Payment_Method' in df_clean.columns:
    
    payment_analysis = df_clean.groupby(
        'Payment_Method'
    ).size().sort_values(ascending=False)

    display(payment_analysis)

    print("Most Popular Payment Method:")
    print(payment_analysis.idxmax())
```

```python
if all(col in df_clean.columns for col in ['Delivery_Mode', 'Delivery_Days']):
    
    delivery_analysis = df_clean.groupby(
        'Delivery_Mode'
    )['Delivery_Days'].mean().sort_values()

    display(delivery_analysis)

    print("Fastest Average Delivery Mode:")
    print(delivery_analysis.idxmin())
```

```python
if all(col in df_clean.columns for col in ['Region', 'Sales']):
    
    plt.figure(figsize=(9, 5))

    sns.barplot(
        data=df_clean,
        x='Region',
        y='Sales',
        estimator='sum',
        errorbar=None
    )

    plt.title('Total Sales by Region')
    plt.xlabel('Region')
    plt.ylabel('Total Sales')

    plt.show()
```

```python
if all(col in df_clean.columns for col in ['Product_Category', 'Profit']):
    
    plt.figure(figsize=(9, 5))

    sns.barplot(
        data=df_clean,
        x='Product_Category',
        y='Profit',
        estimator='sum',
        errorbar=None
    )

    plt.title('Total Profit by Product Category')
    plt.xticks(rotation=30)

    plt.show()
if all(col in df_clean.columns for col in ['Order_Date', 'Sales']):
    
    monthly_sales = df_clean.groupby(
        df_clean['Order_Date'].dt.to_period('M')
    )['Sales'].sum()

    monthly_sales.index = monthly_sales.index.astype(str)

    plt.figure(figsize=(12, 5))

    plt.plot(
        monthly_sales.index,
        monthly_sales.values,
        marker='o'
    )

    plt.title('Monthly Sales Trend')
    plt.xlabel('Month')
    plt.ylabel('Total Sales')

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.show()
if 'Satisfaction_Level' in df_clean.columns:
    
    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=df_clean,
        x='Satisfaction_Level',
        order=['Low', 'Medium', 'High']
    )

    plt.title('Customer Satisfaction Distribution')
    plt.xlabel('Satisfaction Level')
    plt.ylabel('Number of Transactions')

    plt.show()
if all(col in df_clean.columns for col in ['Sales', 'Profit']):
    
    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df_clean,
        x='Sales',
        y='Profit'
    )

    plt.title('Relationship Between Sales and Profit')
    plt.xlabel('Sales')
    plt.ylabel('Profit')

    plt.show()
print("FINAL DATASET INFORMATION")

print("\nDataset Shape:")
print(df_clean.shape)

print("\nData Types:")
print(df_clean.dtypes)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nDuplicate Records:")
print(df_clean.duplicated().sum())

print("\nFirst Five Cleaned Records:")
display(df_clean.head())
```

```python
output_file = "cleaned_retail_customer_sales_dataset_final.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File Name:", output_file)
print("File Location:", os.path.abspath(output_file))
```
